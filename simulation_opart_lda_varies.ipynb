```python
import numpy as np
import matplotlib.pyplot as plt
import opart_lda

np.random.seed(1234)
```

```python
figsize = (7, 1)
```

```python
def plot_chpnt(sequence, chpnts):
    plt.figure(figsize=figsize)
    plt.scatter(np.arange(len(sequence)), sequence, color='black')

    # Add vertical lines at positions in chpnt
    for position in chpnts:
        plt.axvline(x=position + 0.5, color='red', linestyle='--')
    
    plt.xticks([])
    plt.yticks([])
    plt.show()
```

```python
# template vector
values = [1, 2, 1, 2]
segment_length = 40 // len(values)
noise_strength = 0.2

sequence = np.concatenate([
    np.full(segment_length, value) + np.random.normal(0, noise_strength, segment_length) 
    for value in values
])
```

```python
plt.figure(figsize=figsize)
plt.xticks([])
plt.yticks([])
plt.scatter(np.arange(len(sequence)), sequence, color='black')
```

```python
for penalty in [0, 0.1, 0.5, 4, 8]:
    plot_chpnt(sequence, opart_lda.opart(sequence, penalty))
```
